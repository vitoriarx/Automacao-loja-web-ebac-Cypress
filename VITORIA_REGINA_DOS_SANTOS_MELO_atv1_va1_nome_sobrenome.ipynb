## Questão 19) Usando os métodos da biblioteca OpenCV realize a rotação em 45º e uma translação equivalente a 1/4 da da altura da imagem.

```python
# Código aqui
# Usa a imagem aleatória criada anteriormente
imagem = imagem_aleatoria_cinza

altura, largura = imagem.shape

# Rotação de 45 graus
centro = (largura // 2, altura // 2)
matriz_rotacao = cv2.getRotationMatrix2D(centro, 45, 1)

imagem_rotacionada = cv2.warpAffine(
    imagem,
    matriz_rotacao,
    (largura, altura)
)

# Translação de 1/4 da altura
translacao = altura // 4

matriz_translacao = np.float32([
    [1, 0, translacao],
    [0, 1, 0]
])

imagem_final = cv2.warpAffine(
    imagem_rotacionada,
    matriz_translacao,
    (largura, altura)
)

plt.imshow(imagem_final, cmap='gray')
plt.axis('off')
plt.show()
```

## Questão 20) Implemente uma função de translação em uma imagem sem usar a biblioteca OpenCV.

```python
# Código aqui
def transladar_imagem(imagem, deslocamento_x, deslocamento_y):
    altura, largura = imagem.shape[:2]

    imagem_transladada = np.zeros_like(imagem)

    imagem_transladada[
        max(0, deslocamento_y):min(altura, altura + deslocamento_y),
        max(0, deslocamento_x):min(largura, largura + deslocamento_x)
    ] = imagem[
        max(0, -deslocamento_y):min(altura, altura - deslocamento_y),
        max(0, -deslocamento_x):min(largura, largura - deslocamento_x)
    ]

    return imagem_transladada


imagem_transladada = transladar_imagem(
    imagem_aleatoria_cinza,
    32,
    0
)

plt.imshow(imagem_transladada, cmap='gray')
plt.axis('off')
plt.show()
```
